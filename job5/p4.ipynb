```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
from skimage.feature import match_template

# Muat citra target dan template
image = data.coins()
template = image[15:80, 190:260]

# Matching
result = match_template(image, template)
ij = np.unravel_index(np.argmax(result), result.shape)
y, x = ij

# Visualisasi
fig, axes = plt.subplots(1,3, figsize=(12,4))
axes[0].imshow(image, cmap='gray'); axes[0].set_title('Target'); axes[0].axis('off')
axes[1].imshow(template, cmap='gray'); axes[1].set_title('Template'); axes[1].axis('off')
axes[2].imshow(result, cmap='viridis'); axes[2].set_title('Heatmap'); axes[2].plot(x, y, 'r+', markersize=12); axes[2].axis('off')
plt.tight_layout(); plt.show()

# Tampilkan kotak pada lokasi terbaik
h, w = template.shape
fig2, ax2 = plt.subplots(figsize=(6,6))
ax2.imshow(image, cmap='gray'); ax2.axis('off')
rect = plt.Rectangle((x, y), w, h, edgecolor='r', facecolor='none', lw=2)
ax2.add_patch(rect)
plt.show()

print(f"Lokasi terbaik (row, col): ({y}, {x})")
```
